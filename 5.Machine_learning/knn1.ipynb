```python
xx = df.drop(['Symptom_3'], axis=1)
```

```python
yy = df.Symptom_3.values
```

```python
from sklearn.model_selection import train_test_split

xx_train, xx_test, yy_train, yy_test = train_test_split(xx,yy,test_size=0.25)
```

```python
len(xx_train)
```

```python
len(xx_test)
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
sc.fit(xx_train)
xx_train_sc = sc.transform(xx_train)
xx_test_sc = sc.transform(xx_test)
```

```python
xx_train_sc
```

```python
xx_test_sc
```

```python
from sklearn.neighbors import KNeighborsClassifier

model_knn = KNeighborsClassifier(n_neighbors=7)

model_knn.fit(xx_train_sc, yy_train)
```

```python
yy_pred_knn = model_knn.predict(xx_test_sc)
```

```python
yy_pred_knn
```

```python
yy_test
```

```python
from sklearn.metrics import accuracy_score

print(round(accuracy_score(yy_test,yy_pred_knn)*100,2))
```
