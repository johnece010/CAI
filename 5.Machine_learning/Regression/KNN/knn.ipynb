```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("data/cancer_dataset (1).csv")
```

```python
df.dropna(how='any',inplace=True)
```

```python
len(df)
```

```python
df.diagnosis.value_counts()
```

```python
df.diagnosis.value_counts()/len(df)*100
```

```python
X = df.drop(['id','diagnosis'],axis=1)
```

```python
Y = df.diagnosis.values
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.25)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
X_train_sc = sc.fit_transform(X_train)
X_test_sc = sc.fit_transform(X_test)
```

```python
X_train_sc
```

```python
X_test_sc
```

```python
from sklearn.neighbors import KNeighborsClassifier

model_knn = KNeighborsClassifier(n_neighbors=7)

model_knn.fit(X_train_sc, Y_train)
```

```python
Y_pred_knn = model_knn.predict(X_test_sc)
```

```python
Y_pred_knn
```

```python
Y_test
```

```python
from sklearn.metrics import accuracy_score
scores = []

for i in range(1, 21):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train_sc,Y_train)
    Y_pred_knn = knn.predict(X_test_sc)
    scores.append(accuracy_score(Y_test, Y_pred_knn))
```

```python
plt.plot(range(1, 21), scores)
plt.show()
```

```python
from sklearn.metrics import accuracy_score

print(round(accuracy_score(Y_test,Y_pred_knn)*100,2))
```
